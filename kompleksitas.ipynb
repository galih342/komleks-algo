```python
# ==========================================
# PROGRAM DIVIDE AND CONQUER
# MERGE SORT
# ==========================================

# Fungsi untuk menggabungkan dua bagian array
def merge(kiri, kanan):

    hasil = []

    i = 0
    j = 0

    # Membandingkan elemen dari kedua bagian
    while i < len(kiri) and j < len(kanan):

        if kiri[i] <= kanan[j]:
            hasil.append(kiri[i])
            i += 1

        else:
            hasil.append(kanan[j])
            j += 1

    # Memasukkan sisa elemen dari bagian kiri
    while i < len(kiri):
        hasil.append(kiri[i])
        i += 1

    # Memasukkan sisa elemen dari bagian kanan
    while j < len(kanan):
        hasil.append(kanan[j])
        j += 1

    return hasil


# Fungsi utama Merge Sort
def merge_sort(data, tingkat=0):

    # Kondisi berhenti (base case)
    if len(data) <= 1:
        return data

    # Membagi data menjadi dua bagian
    tengah = len(data) // 2

    kiri = data[:tengah]
    kanan = data[tengah:]

    indentasi = "  " * tingkat

    print(indentasi + "Membagi data:", data)
    print(indentasi + "Bagian kiri:", kiri)
    print(indentasi + "Bagian kanan:", kanan)

    # Memanggil fungsi secara rekursif
    kiri_urut = merge_sort(kiri, tingkat + 1)
    kanan_urut = merge_sort(kanan, tingkat + 1)

    # Menggabungkan kedua bagian yang sudah terurut
    hasil = merge(kiri_urut, kanan_urut)

    print(indentasi + "Hasil penggabungan:", hasil)

    return hasil


# ==========================================
# PROGRAM UTAMA
# ==========================================

print("================================")
print("PROGRAM MERGE SORT")
print("DIVIDE AND CONQUER")
print("================================")

# Data awal
data = [38, 27, 43, 3, 9, 82, 10]

print("\nData sebelum diurutkan:")
print(data)

# Menjalankan algoritma Merge Sort
hasil = merge_sort(data)

print("\n================================")
print("HASIL AKHIR")
print("================================")

print("Data sebelum diurutkan:", data)
print("Data setelah diurutkan:", hasil)
```

```python
# ==========================================
# PROGRAM GREEDY ALGORITHM
# FRACTIONAL KNAPSACK
# ==========================================

def fractional_knapsack(barang, kapasitas):

    # Menghitung rasio nilai keuntungan terhadap berat
    print("Menghitung rasio keuntungan setiap barang:")

    for item in barang:
        item["rasio"] = item["nilai"] / item["berat"]

        print(
            item["nama"],
            "-> Rasio:",
            item["rasio"]
        )

    # Mengurutkan barang berdasarkan rasio tertinggi
    barang_urut = sorted(
        barang,
        key=lambda x: x["rasio"],
        reverse=True
    )

    print("\nUrutan barang berdasarkan rasio tertinggi:")

    for item in barang_urut:
        print(
            item["nama"],
            "| Berat:", item["berat"],
            "| Nilai:", item["nilai"],
            "| Rasio:", item["rasio"]
        )

    # Variabel untuk menyimpan hasil
    total_nilai = 0
    total_berat = 0

    print("\nProses pemilihan barang:")

    # Memilih barang berdasarkan urutan rasio
    for item in barang_urut:

        if kapasitas <= 0:
            break

        # Jika berat barang masih mencukupi kapasitas
        if item["berat"] <= kapasitas:

            kapasitas -= item["berat"]

            total_berat += item["berat"]
            total_nilai += item["nilai"]

            print(
                item["nama"],
                "diambil seluruhnya"
            )

            print("Berat diambil:", item["berat"])
            print("Nilai diperoleh:", item["nilai"])

        # Jika kapasitas tidak mencukupi
        else:

            # Menghitung bagian barang yang dapat diambil
            bagian = kapasitas / item["berat"]

            berat_diambil = kapasitas
            nilai_diperoleh = item["nilai"] * bagian

            total_berat += berat_diambil
            total_nilai += nilai_diperoleh

            print(
                item["nama"],
                "diambil sebagian"
            )

            print("Persentase diambil:", bagian * 100, "%")
            print("Berat diambil:", berat_diambil)
            print("Nilai diperoleh:", nilai_diperoleh)

            kapasitas = 0

    return total_berat, total_nilai


# ==========================================
# PROGRAM UTAMA
# ==========================================

print("================================")
print("PROGRAM FRACTIONAL KNAPSACK")
print("GREEDY ALGORITHM")
print("================================")

# Data barang
barang = [
    {"nama": "Barang A", "berat": 10, "nilai": 60},
    {"nama": "Barang B", "berat": 20, "nilai": 100},
    {"nama": "Barang C", "berat": 30, "nilai": 120}
]

# Kapasitas maksimum tas
kapasitas = 50

print("\nKapasitas maksimum tas:", kapasitas)

# Menampilkan data barang
print("\nData barang:")

for item in barang:
    print(
        item["nama"],
        "| Berat:", item["berat"],
        "| Nilai:", item["nilai"]
    )

# Menjalankan algoritma
berat, keuntungan = fractional_knapsack(
    barang,
    kapasitas
)

# Menampilkan hasil akhir
print("\n================================")
print("HASIL AKHIR")
print("================================")

print("Total berat barang:", berat)
print("Total keuntungan maksimum:", keuntungan)
```

```python
# ==========================================
# PROGRAM DYNAMIC PROGRAMMING
# COIN CHANGE
# ==========================================

def coin_change(koin, target):

    # Membuat array untuk menyimpan hasil minimum koin
    # Nilai awal dibuat tak terhingga
    dp = [float('inf')] * (target + 1)

    # Array untuk menyimpan koin yang dipilih
    pilihan = [None] * (target + 1)

    # Kondisi awal:
    # Untuk membentuk nilai 0 tidak membutuhkan koin
    dp[0] = 0

    print("================================")
    print("PROSES DYNAMIC PROGRAMMING")
    print("================================")

    # Menghitung nilai minimum untuk setiap target
    for jumlah in range(1, target + 1):

        print("\nMenghitung nilai:", jumlah)

        # Memeriksa semua jenis koin
        for nilai_koin in koin:

            # Koin hanya dapat digunakan jika nilainya
            # tidak melebihi jumlah yang sedang dihitung
            if nilai_koin <= jumlah:

                # Menghitung kemungkinan jumlah koin baru
                kandidat = dp[jumlah - nilai_koin] + 1

                print(
                    "Menggunakan koin",
                    nilai_koin,
                    "-> kandidat:",
                    kandidat
                )

                # Memilih jumlah koin yang paling sedikit
                if kandidat < dp[jumlah]:

                    dp[jumlah] = kandidat

                    # Menyimpan koin yang dipilih
                    pilihan[jumlah] = nilai_koin

        print(
            "Minimum koin untuk",
            jumlah,
            "=",
            dp[jumlah]
        )

    # Jika target tidak dapat dibentuk
    if dp[target] == float('inf'):
        return -1, [], dp

    # Menyusun kembali kombinasi koin yang digunakan
    hasil_koin = []
    sisa = target

    while sisa > 0:

        koin_dipilih = pilihan[sisa]

        hasil_koin.append(koin_dipilih)

        sisa -= koin_dipilih

    return dp[target], hasil_koin, dp


# ==========================================
# PROGRAM UTAMA
# ==========================================

print("================================")
print("PROGRAM COIN CHANGE")
print("DYNAMIC PROGRAMMING")
print("================================")

# Jenis koin yang tersedia
koin = [1, 3, 4]

# Target nilai yang ingin dibentuk
target = 6

print("\nJenis koin:", koin)
print("Nilai target:", target)

# Menjalankan algoritma Coin Change
jumlah_minimum, kombinasi, tabel = coin_change(
    koin,
    target
)

# Menampilkan hasil akhir
print("\n================================")
print("HASIL AKHIR")
print("================================")

if jumlah_minimum == -1:

    print("Target tidak dapat dibentuk.")

else:

    print("Jumlah minimum koin:", jumlah_minimum)
    print("Kombinasi koin:", kombinasi)
    print("Total nilai koin:", sum(kombinasi))

# Menampilkan tabel DP
print("\n================================")
print("TABEL DYNAMIC PROGRAMMING")
print("================================")

for i in range(len(tabel)):
    print("Nilai:", i, "| Minimum koin:", tabel[i])
```
